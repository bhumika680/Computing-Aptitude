```python
# Student Record Management & Analytics
# Part 1

import pandas as pd
import numpy as np


# -------------------------------
# 1. Student Class
# -------------------------------

class Student:

    def __init__(self, student_id, name, math, python, dbms,
                 computer_networks, english):
        self.student_id = student_id
        self.name = name
        self.math = math
        self.python = python
        self.dbms = dbms
        self.computer_networks = computer_networks
        self.english = english

    def total_marks(self):
        return (
            self.math +
            self.python +
            self.dbms +
            self.computer_networks +
            self.english
        )

    def average_marks(self):
        return self.total_marks() / 5


# -------------------------------
# 2. Load Dataset
# -------------------------------

file = "student_record_management_60_students.csv"

df = pd.read_csv(file)

print("Student Record Management & Analytics")
print("------------------------------------")

print("\nFirst 5 Records:")
print(df.head())


# -------------------------------
# 3. Basic Information
# -------------------------------

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())


# -------------------------------
# 4. Basic Statistics
# -------------------------------

subjects = [
    "Math",
    "Python",
    "DBMS",
    "Computer_Networks",
    "English"
]

print("\nAverage Marks in Each Subject:")

for subject in subjects:
    print(subject, ":", round(df[subject].mean(), 2))


# -------------------------------
# 5. Overall Statistics
# -------------------------------

print("\nOverall Statistics")

print("Highest Average:",
      df["Average_Marks"].max())

print("Lowest Average:",
      df["Average_Marks"].min())

print("Class Average:",
      round(df["Average_Marks"].mean(), 2))

print("Average Attendance:",
      round(df["Attendance_Percent"].mean(), 2))


# -------------------------------
# 6. Grade Analysis
# -------------------------------

print("\nNumber of Students in Each Grade:")
print(df["Grade"].value_counts())


# -------------------------------
# 7. Result Analysis
# -------------------------------

print("\nPass and Fail Students:")
print(df["Result"].value_counts())


# -------------------------------
# 8. Top Students
# -------------------------------

top_students = df.sort_values(
    by="Average_Marks",
    ascending=False
).head(5)

print("\nTop 5 Students:")
print(
    top_students[
        ["Student_ID", "Name", "Average_Marks", "Grade"]
    ]
)
```

```python
# Student Record Management & Analytics
# Part 2

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# -------------------------------
# 1. Load Dataset
# -------------------------------

file = "student_record_management_60_students.csv"

df = pd.read_csv(file)


# -------------------------------
# 2. Student Class
# -------------------------------

class Student:

    def __init__(self, student_id, name, math, python, dbms,
                 computer_networks, english):
        self.student_id = student_id
        self.name = name
        self.math = math
        self.python = python
        self.dbms = dbms
        self.computer_networks = computer_networks
        self.english = english

    def total_marks(self):
        return (
            self.math + self.python + self.dbms +
            self.computer_networks + self.english
        )

    def average_marks(self):
        return self.total_marks() / 5


# -------------------------------
# 3. Create Student Objects
# -------------------------------

students = []

for _, row in df.iterrows():

    student = Student(
        row["Student_ID"],
        row["Name"],
        row["Math"],
        row["Python"],
        row["DBMS"],
        row["Computer_Networks"],
        row["English"]
    )

    students.append(student)


print("Total Student Objects Created:", len(students))


# -------------------------------
# 4. Example Object
# -------------------------------

student = students[0]

print("\nExample Student Object")
print("Student ID:", student.student_id)
print("Name:", student.name)
print("Total Marks:", student.total_marks())
print("Average Marks:", round(student.average_marks(), 2))


# -------------------------------
# 5. NumPy Analysis
# -------------------------------

marks = df[
    ["Math", "Python", "DBMS",
     "Computer_Networks", "English"]
].values

print("\nNumPy Analysis")

print("Overall Mean Marks:",
      round(np.mean(marks), 2))

print("Highest Mark:",
      np.max(marks))

print("Lowest Mark:",
      np.min(marks))

print("Standard Deviation:",
      round(np.std(marks), 2))


# -------------------------------
# 6. Subject Average
# -------------------------------

subject_average = df[
    ["Math", "Python", "DBMS",
     "Computer_Networks", "English"]
].mean()

print("\nSubject Average:")
print(subject_average)


# -------------------------------
# 7. Highest Performing Subject
# -------------------------------

best_subject = subject_average.idxmax()
lowest_subject = subject_average.idxmin()

print("\nBest Performing Subject:",
      best_subject)

print("Lowest Performing Subject:",
      lowest_subject)


# -------------------------------
# 8. Attendance Analysis
# -------------------------------

high_attendance = df[
    df["Attendance_Percent"] >= 85
]

low_attendance = df[
    df["Attendance_Percent"] < 85
]

print("\nStudents with Attendance >= 85:",
      len(high_attendance))

print("Average marks of high attendance students:",
      round(high_attendance["Average_Marks"].mean(), 2))

print("Average marks of lower attendance students:",
      round(low_attendance["Average_Marks"].mean(), 2))


# -------------------------------
# 9. Semester Analysis
# -------------------------------

semester_average = df.groupby(
    "Semester"
)["Average_Marks"].mean()

print("\nAverage Marks by Semester:")
print(semester_average.round(2))


# -------------------------------
# 10. Gender Analysis
# -------------------------------

gender_average = df.groupby(
    "Gender"
)["Average_Marks"].mean()

print("\nAverage Marks by Gender:")
print(gender_average.round(2))


# ==================================================
# VISUALIZATIONS
# ==================================================


# -------------------------------
# 11. Subject Average Bar Chart
# -------------------------------

plt.figure(figsize=(8, 5))

subject_average.plot(kind="bar")

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# -------------------------------
# 12. Grade Distribution
# -------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Grade"
)

plt.title("Student Grade Distribution")
plt.xlabel("Grade")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# -------------------------------
# 13. Attendance vs Marks
# -------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance_Percent",
    y="Average_Marks"
)

plt.title("Attendance vs Average Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")

plt.tight_layout()
plt.show()


# -------------------------------
# 14. Student Performance
# -------------------------------

top_10 = df.sort_values(
    "Average_Marks",
    ascending=False
).head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=top_10,
    x="Name",
    y="Average_Marks"
)

plt.title("Top 10 Students")
plt.xlabel("Student")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# -------------------------------
# 15. Marks Distribution
# -------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    df["Average_Marks"],
    bins=8
)

plt.title("Distribution of Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# -------------------------------
# 16. Subject Performance
# -------------------------------

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df[
        ["Math", "Python", "DBMS",
         "Computer_Networks", "English"]
    ]
)

plt.title("Subject-wise Marks Distribution")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# -------------------------------
# 17. Semester Performance
# -------------------------------

plt.figure(figsize=(7, 5))

semester_average.plot(
    kind="bar"
)

plt.title("Average Marks by Semester")
plt.xlabel("Semester")
plt.ylabel("Average Marks")

plt.tight_layout()
plt.show()


# -------------------------------
# 18. Gender Performance
# -------------------------------

plt.figure(figsize=(7, 5))

gender_average.plot(
    kind="bar"
)

plt.title("Average Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")

plt.tight_layout()
plt.show()


# -------------------------------
# 19. Correlation Heatmap
# -------------------------------

plt.figure(figsize=(9, 6))

numeric_columns = [
    "Math",
    "Python",
    "DBMS",
    "Computer_Networks",
    "English",
    "Attendance_Percent",
    "Average_Marks"
]

sns.heatmap(
    df[numeric_columns].corr(),
    annot=True,
    fmt=".2f"
)

plt.title("Student Performance Correlation")

plt.tight_layout()
plt.show()


# -------------------------------
# 20. Final Insights
# -------------------------------

print("\n========== PROJECT INSIGHTS ==========")

print(
    "1. Best performing subject:",
    best_subject
)

print(
    "2. Lowest performing subject:",
    lowest_subject
)

print(
    "3. Class average:",
    round(df["Average_Marks"].mean(), 2)
)

print(
    "4. Average attendance:",
    round(df["Attendance_Percent"].mean(), 2)
)

print(
    "5. Number of students:",
    len(df)
)

print(
    "6. Highest average marks:",
    df["Average_Marks"].max()
)

print(
    "7. Lowest average marks:",
    df["Average_Marks"].min()
)

print(
    "8. Most common grade:",
    df["Grade"].mode()[0]
)

print(
    "9. Most common result:",
    df["Result"].mode()[0]
)

print(
    "10. Highest attendance:",
    df["Attendance_Percent"].max()
)

print("\n========== PROJECT COMPLETED ==========")
```
